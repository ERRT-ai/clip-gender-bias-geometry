## Research question

When occupational gender association gaps are measured in CLIP, which evaluation choice changes the measured gap the most?

The experiment varies three factors: evaluation-pool gender composition, prompt formulation, and embedding geometry. The primary outcome is `|mean(similarity_male) - mean(similarity_female)|`.

## Headline results

| Factor | Partial η² |
| --- | ---: |
| Embedding method | **0.644729** |
| Prompt type | 0.066000 |
| Data setting | 0.006459 |

Matched across 180 conditions, the mean absolute gap is **0.002056** after gender-direction removal, compared with **0.010381** under standard cosine similarity and **0.022689** after mean-centering.


## Mechanism result

Under the balanced evaluation pool and standard cosine similarity, the projection of occupation text embeddings onto the estimated image-derived gender direction is almost perfectly aligned with the signed gender association gap: **Pearson r = 0.999851** (N=60).

This supports a representation-geometry interpretation of the measured signal, while not implying that removing one linear direction makes CLIP generally fair.
